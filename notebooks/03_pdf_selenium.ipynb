```python
# FUNCTIONS

async def get_job_description(job_url):
    async with BrowserManager(headless=False) as browser:
        
        # You'll need a session file with your LinkedIn login
        await browser.load_session("session.json")
        
        scraper = JobSearchScraper(browser.page)
        # This returns structured job data including description
        job = await scraper.scrape_job_from_url(job_url)
        
        return {
            "title": job.title,
            "company": job.company,
            "description": job.description,
            "location": job.location
        }

async def create_session():
    async with BrowserManager(headless=False) as browser:
        await browser.page.goto("https://www.linkedin.com/login")
        print("Please log in manually in the browser...")
        await wait_for_manual_login(browser.page, timeout=300)
        await browser.save_session("session.json")  # Reuse this
```
